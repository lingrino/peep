# Sales analysis

A notebook rendered as a document: Markdown cells, highlighted code, and outputs (text, tables, errors, images).

```python
import statistics
sales = [12, 19, 7, 25, 16, 22, 9, 28]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug']
```

## Summary statistics

```python
for name, fn in [('mean', statistics.mean), ('median', statistics.median), ('stdev', statistics.stdev)]:
    print(f'{name:<6} {fn(sales):.2f}'.rstrip('0'))
```

```python
{'best': months[sales.index(max(sales))], 'worst': months[sales.index(min(sales))]}
```

## Chart

Image outputs are embedded in the notebook and drawn inline.

```python
import matplotlib.pyplot as plt
plt.bar(months, sales)
plt.show()
```
